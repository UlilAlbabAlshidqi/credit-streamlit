# Kategorik Ordinal dengan Target

```python
from scipy.stats import spearmanr

kategorik_ordinal = [
    'PAY_0', 'PAY_2', 'PAY_3',
    'PAY_4', 'PAY_5', 'PAY_6'
]

for var in kategorik_ordinal:
    rho, p_spearman = spearmanr(
        *df_credit_data[[var, 'default.payment.next.month']] # artinya apa
        .dropna()
        .to_numpy().T
    )

    print(f"\n{var}")
    print(f"Spearman rho     : {rho:.4f}")  #rho itu arah dan kekuatan hubungan monotonik
    print(f"Spearman p-value : {p_spearman:.4g}") # p value artinya menguji hipotesisi statissik (<0.05 tolak H0 terima H1)
```

```python
# semua variabel menolak h0 dan menerima H1 (p <0.05) artinya terdapat korelasi monotonik antara PAY dengan target
# sayangnya kekuatan korelasinya lemah hingga sangat lemah (0.14-0.29)
```

```python
# Feature ordinal terhadap target biner: Mann–Whitney U karena target hanya berupa dua kelompok independen, kalau misal targetnya tiga maka bisa pakai uji kruskall wallis dan jika p value< 0.05 bisa lanjut post hucc dunn test

from scipy.stats import mannwhitneyu

for col in kategorik_ordinal:
    group_0 = df_credit_data.loc[
        df_credit_data['default.payment.next.month'] == 0, col
    ].dropna()

    group_1 = df_credit_data.loc[
        df_credit_data['default.payment.next.month'] == 1, col
    ].dropna()

    stat, p_value = mannwhitneyu(
        group_0,
        group_1,
        alternative='two-sided'
    )

    print(f'{col}: U={stat:.2f}, p-value={p_value:.3e}')
```

```python
# Hasil ini menunjukkan bahwa keenam fitur status keterlambatan pembayaran (PAY_0–PAY_6) memiliki perbedaan distribusi yang signifikan antara nasabah default dan non-default berdasarkan uji Mann–Whitney U.
```

# Uji Asosiasi Kategorik Nominal dgn Target

```python
# target default.payment.next.month berisi 0 dan 1. Ketika menguji hubungannya dengan SEX, EDUCATION, atau MARRIAGE, gunakan Chi-square karena keduanya merupakan variabel kategorikal.
```

```python

from scipy.stats import chi2_contingency
import pandas as pd

# Feature kategorik nominal terhadap target biner
kategorik_nominal = ['SEX', 'EDUCATION', 'MARRIAGE']
target = 'default.payment.next.month'

hasil = []

for col in kategorik_nominal:
    data = df_credit_data[[col, target]].dropna()

    # Tabel kontingensi feature × target
    tabel = pd.crosstab(data[col], data[target])

    if tabel.shape[0] < 2 or tabel.shape[1] < 2:
        continue

    chi2, p_value, _, expected = chi2_contingency(tabel)

    # Ukuran efek Cramér's V
    n = tabel.to_numpy().sum()
    r, k = tabel.shape
    cramers_v = (chi2 / (n * min(r - 1, k - 1))) ** 0.5

    hasil.append({
        'Variabel': col,
        'Jumlah Kategori': data[col].nunique(),
        'Uji': 'Chi-square',
        'Statistic': chi2,
        'p-value': p_value,
        "Cramer's V": cramers_v,
        'Kesimpulan': (
            'Signifikan' if p_value < 0.05
            else 'Tidak signifikan'
        )
    })

hasil_nominal = pd.DataFrame(hasil)
hasil_nominal
```

```python
# 1. p-value — Ketiga variabel memiliki p-value di bawah 0,05, sehingga terdapat bukti statistik adanya hubungan dengan default.payment.next.month.

#2. Cramér’s V digunakan untuk mengukur kekuatan hubungan antara dua variabel kategorikal. Berdasarkan hasil pengujian, variabel SEX, EDUCATION, dan MARRIAGE terhadap target default.payment.next.month seluruhnya menunjukkan hubungan dengan kekuatan asosiasi yang sangat lemah.
```

#Asosiasi antar Kategorik baik Nominal dan Ordinal

# --> Chi-Square dan Cramér's V
berfungsi untuk menganalisis asosiasi antara dua variabel kategorikal.

Chi-Square menguji apakah terdapat asosiasi secara statistik, sedangkan Cramér's V mengukur kekuatan asosiasi tersebut.




```python
from scipy.stats import chi2_contingency
import pandas as pd
import numpy as np

kat_cols = kategorik_ordinal + kategorik_nominal

hasil = []

for i in range(len(kat_cols)):
    for j in range(i + 1, len(kat_cols)):
        col1 = kat_cols[i]
        col2 = kat_cols[j]

        data = df_credit_data[[col1, col2]].dropna()

        tabel = pd.crosstab(
            data[col1],
            data[col2]
        )

        if tabel.shape[0] < 2 or tabel.shape[1] < 2:
            continue

        chi2, p_value, dof, expected = chi2_contingency(tabel)

        n = tabel.values.sum()
        r, k = tabel.shape

        cramers_v = np.sqrt(
            chi2 / (n * min(r - 1, k - 1))
        )

        hasil.append({
            'Variabel 1': col1,
            'Variabel 2': col2,
            'Chi-Square': chi2,
            'p-value': p_value,
            "Cramer's V": cramers_v
        })

hasil_chi = pd.DataFrame(hasil)

hasil_chi
```

```python
# Analisis Chi-Square menunjukkan adanya asosiasi statistik antarvariabel kategorik.
#Asosiasi terkuat ditemukan di antara variabel status pembayaran (PAY_0–PAY_6), dengan Cramer's V sekitar 0,47–0,71 untuk pasangan yang ditampilkan.
# Sebaliknya, asosiasi antara variabel demografis dan status pembayaran cenderung lemah. Temuan ini mengindikasikan bahwa pola keterlambatan pembayaran antar bulan saling berkaitan
```

```python
# Preprocessing

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    RobustScaler,
    OneHotEncoder,
    FunctionTransformer,
    PowerTransformer
)

# 1. Pipeline numerik biasa
numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# 2. Pipeline numerik dengan log1p + RobustScaler
#Function trasnformer ini python menjalankan fungsin np.log1p ke setiap nilai pada kolom yang telah didefinsikan

numeric_log_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('log', FunctionTransformer(
        np.log1p,
        feature_names_out='one-to-one'
    )),
    ('scaler', RobustScaler())
])

# 3. Pipeline numerik dengan Yeo-Johnson + RobustScaler
# Power Transformer scikit-learn menerapkan metode Yeo-Johnson pada kolom yang telah didefinisikan

numeric_yeo_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('yeo_johnson', PowerTransformer(
        method='yeo-johnson',
        standardize=False
    )),
    ('scaler', RobustScaler())
])

# 4. Fungsi konversi fitur kategorikal menjadi string
def categorical_to_string(X):
    X = pd.DataFrame(X).copy()
    return X.map(
        lambda value: str(value) if pd.notna(value) else np.nan
    ).to_numpy()

# 5. Pipeline fitur ordinal
kategorik_ordinal_pipeline = Pipeline([
    ('to_string', FunctionTransformer(
        categorical_to_string,
        feature_names_out='one-to-one'
    )),
    ('imputer', SimpleImputer(
        strategy='constant',
        fill_value='KOSONG'
    )),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    ))
])

# 6. Pipeline fitur nominal
kategorik_nominal_pipeline = Pipeline([
    ('to_string', FunctionTransformer(
        categorical_to_string,
        feature_names_out='one-to-one'
    )),
    ('imputer', SimpleImputer(
        strategy='constant',
        fill_value='KOSONG'
    )),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    ))
])

# 7. Gabungkan seluruh preprocessing
preprocessor = ColumnTransformer([
    ('numeric', numeric_pipeline, numeric),
    ('numeric_log', numeric_log_pipeline, numeric_log),
    ('numeric_yeo', numeric_yeo_pipeline, numeric_yeo_johnson),
    ('kategorik_ordinal', kategorik_ordinal_pipeline, kategorik_ordinal),
    ('kategorik_nominal', kategorik_nominal_pipeline, kategorik_nominal)
])
```

```python
# Definisikan Model non Boosting

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models_non_boosting = {
    'Baseline': DummyClassifier(
        strategy='most_frequent'
    ),

    'Logistic Regression': LogisticRegression(
        class_weight='balanced',
        max_iter=1000,
        random_state=42
    ),

    'KNN': KNeighborsClassifier(
        n_neighbors=5,
        weights='distance'
    ),

    'Decision Tree': DecisionTreeClassifier(
        class_weight='balanced',
        random_state=42
    ),

    'Random Forest': RandomForestClassifier(
        n_estimators=200,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    )
}
```

```python
# Gabungkan preprocessing dan model dalam pipeline agar imputasi dan preprocessing di-fit ulang pada setiap fold CV,
#untuk mencegah data leakage pada data validasi

from sklearn.base import clone

pipelines_non_boosting = {
    name: Pipeline([
        ('preprocessor', clone(preprocessor)),
        ('model', model)
    ])
    for name, model in models_non_boosting.items()
}
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    'Accuracy': 'accuracy',
    'Precision': 'precision',
    'Recall': 'recall',
    'F1-Score': 'f1',
    'ROC-AUC': 'roc_auc',
    'PR-AUC': 'average_precision'
}
```

```python
results_non_boosting = []

for name, pipeline in pipelines_non_boosting.items():

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=1
    )

    row = {'Model': name}

    for metric in scoring:
        row[metric] = scores[f'test_{metric}'].mean()

    results_non_boosting.append(row)
```

```python
results_df_cv_non_boosting = (
    pd.DataFrame(results_non_boosting)
    .sort_values('F1-Score', ascending=False)
    .reset_index(drop=True)
)

display(results_df_cv_non_boosting.round(4))
```

```python
# Definisikan Model Boosting
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier

# Rasio kelas untuk menangani ketidakseimbangan target
scale_pos = (y_train == 0).sum() / (y_train == 1).sum()

models_boosting = {
    'XGBoost': XGBClassifier(
        n_estimators=200,
        learning_rate=0.1,
        max_depth=6,
        scale_pos_weight=scale_pos,
        eval_metric='logloss',
        random_state=42,
        n_jobs=-1
    ),

    'CatBoost': CatBoostClassifier(
        iterations=200,
        learning_rate=0.1,
        depth=6,
        auto_class_weights='Balanced',
        random_seed=42,
        verbose=0,
        allow_writing_files=False
    ),

    'LightGBM': LGBMClassifier(
        n_estimators=200,
        learning_rate=0.1,
        max_depth=-1,
        class_weight='balanced',
        random_state=42,
        verbosity=-1,
        n_jobs=-1
    )
}
```

```python
# Integrasikan preprocessing ke dalam pipeline CV
# agar preprocessing di-fit secara terpisah pada setiap fold

from sklearn.base import clone
from sklearn.pipeline import Pipeline

pipelines_boosting = {
    name: Pipeline([
        ('preprocessor', clone(preprocessor)),
        ('model', model)
    ])
    for name, model in models_boosting.items()
}
```

```python
# Evaluasi model boosting menggunakan 5-fold CV

results_boosting = []

for name, pipeline in pipelines_boosting.items():

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=1
    )

    row = {'Model': name}

    for metric in scoring:
        row[metric] = scores[f'test_{metric}'].mean()

    results_boosting.append(row)
```

```python
# Urutkan hasil berdasarkan F1-Score tertinggi

results_df_cv_boosting = (
    pd.DataFrame(results_boosting)
    .sort_values('F1-Score', ascending=False)
    .reset_index(drop=True)
)

display(results_df_cv_boosting.round(4))
```

```python
# Parameter tunning

from sklearn.model_selection import RandomizedSearchCV

catboost_pipeline = Pipeline([
    ('preprocessor', clone(preprocessor)),
    ('model', CatBoostClassifier(
        auto_class_weights='Balanced',
        random_seed=42,
        verbose=0,
        allow_writing_files=False
    ))
])

param_distributions = {
    'model__iterations': [200, 300, 500, 700, 1000],
    'model__learning_rate': [0.01, 0.03, 0.05, 0.1, 0.15],
    'model__depth': [4, 5, 6, 7, 8, 9, 10],
    'model__l2_leaf_reg': [1, 3, 5, 7, 9, 11],
    'model__random_strength': [0, 0.5, 1, 2, 5],
    'model__border_count': [32, 64, 128, 254]
}
```

```python
#Refit pada seluruh data training

random_search = RandomizedSearchCV(
    estimator=catboost_pipeline,
    param_distributions=param_distributions,
    n_iter=30,
    scoring='f1',
    cv=cv,
    refit=True,
    random_state=42,
    n_jobs=1,
    verbose=1,
    return_train_score=True
)

random_search.fit(X_train, y_train)

print("Best parameters:")
print(random_search.best_params_)

print("Best CV F1-score:")
print(round(random_search.best_score_, 4))
```
